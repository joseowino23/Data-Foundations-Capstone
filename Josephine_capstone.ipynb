**Filling in missing values**

```python
df['total_sales'] = df['total_sales'].fillna(df['unit_price'] * df['quantity'])
```

```python
df['profit'] = df['profit'].fillna(df['total_sales'] - df['cost_price'])
```

```python
df['customer_satisfaction'] = df['customer_satisfaction'].fillna(df['customer_satisfaction'].median())
```

```python
df.isnull().sum()
```

**Convert date to datetime from string**

```python
df['date'] = pd.to_datetime(df['date'],format='mixed', errors='coerce')
```

**Q2.Within electronics, which customer segment has the highest average profit per order?**

```python
Electronics= df[df["product_category"]== "Electronics"]

segment_profit = Electronics.groupby("branch")["profit"].mean()

highest_segment=segment_profit.max()
highest_average_profit=segment_profit.max()

print("highest profit segment:", highest_segment)
print("average profit per order:",highest_average_profit)
```
